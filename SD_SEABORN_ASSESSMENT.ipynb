```python
#Step 1 — Build with AI

#Prompt Used :Create a synthetic dataset with at least 5 numeric columns and 1 categorical column using NumPy and Pandas. Create a Seaborn pairplot colored by the categorical column and save it as a PNG. Also create a Pearson correlation heatmap with annotations using a diverging color palette. Save the heatmap as a separate PNG file with 200 DPI.
#AI-Generated Code
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate synthetic dataset
np.random.seed(42)

n = 250

df = pd.DataFrame({
    "order_value": np.random.uniform(100, 800, n),
    "distance_km": np.random.uniform(1, 20, n),
    "delivery_time_mins": np.random.normal(30, 8, n),
    "rating": np.random.uniform(1, 5, n),
    "discount_pct": np.random.uniform(0, 30, n),
    "restaurant_type": np.random.choice(
        ["Budget", "Regular", "Premium"],
        n
    )
})

print("Dataset Created Successfully")
print(df.head())

# Pairplot
pairplot = sns.pairplot(
    df,
    vars=[
        "order_value",
        "distance_km",
        "delivery_time_mins",
        "rating",
        "discount_pct"
    ],
    hue="restaurant_type"
)

plt.savefig(
    "section_d_pairplot.png",
    dpi=200,
    bbox_inches="tight"
)

plt.close()

# Correlation Heatmap
numeric_columns = df.select_dtypes(
    include=np.number
).columns

correlation = df[numeric_columns].corr()

plt.figure(figsize=(5,5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Pearson Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "section_d_correlation_heatmap.png",
    dpi=200,
    bbox_inches="tight"
)

plt.close()

print("Pairplot saved as section_d_pairplot.png")
print("Heatmap saved as section_d_correlation_heatmap.png")
```
